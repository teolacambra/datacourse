```python
# functions, map, filter, and lambda

# map function
def universe(title: str):
    return title.lower().strip()
universe("Hello my Name is MatThew")
columns = ["Material", "Tensile Strength", "Yield Strength"]
loweredcols = list(map(universe, columns))
print(loweredcols)

# make universe into a lambda expression
def universe(title): return title.lower().strip()
# or
lambda var: var.lower().strip()
```
